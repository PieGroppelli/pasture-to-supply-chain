# From satellite to supply chain
Pasture health → grass growth → cows → milk → supply-chain impact, using Sentinel-2, ERA5-Land and Google AlphaEarth embeddings in Google Earth Engine.

Run the cells in order. You need a free Google Earth Engine project ID (https://code.earthengine.google.com/register).

```python
# 1. Get the code
!git clone https://github.com/PieGroppelli/pasture-to-supply-chain.git
%cd pasture-to-supply-chain
!pip install -q -r requirements.txt
```

```python
# 2. Settings
PROJECT_ID = 'your-earth-engine-project-id'
TARGET_YEAR = 2018    # the 2018 drought in Denmark
BASELINE_YEAR = 2019  # a normal year to compare against
AREAS = {             # one grassland parcel is picked automatically near each location
    'Viborg area': (9.383, 56.430),
    'Videbaek area': (8.623, 56.081),
    'Troldhede area': (8.893, 55.991),
}
FIELDS = 'fields/pastures.geojson'
```

```python
# 2b. Find real grassland parcels (Dynamic World + ESA WorldCover)
from pasture_supply_chain import satellite
satellite.init(PROJECT_ID)
satellite.find_grassland_fields(AREAS, (TARGET_YEAR, BASELINE_YEAR), FIELDS)
```

```python
# 3. Run the full pipeline
from pasture_supply_chain.run import run
table, impact, monthly = run(PROJECT_ID, FIELDS, TARGET_YEAR, BASELINE_YEAR, out_dir='outputs')
```

```python
# 4. Show the results
from IPython.display import Image, display
for f in ['field_maps.png', 'ndvi_curves.png', 'supply_chain.png']:
    display(Image(f'outputs/{f}'))
monthly.round(2)

# Download charts and tables
!zip -qr outputs.zip outputs fields/pastures.geojson
from google.colab import files
files.download('outputs.zip')
```

## Change the assumptions
All non-measured parameters are in `pasture_supply_chain/assumptions.py`. Example: test a feed price of 0.35 €/kg DM.

```python
from pasture_supply_chain import Assumptions
a = Assumptions(replacement_feed_eur_per_kg_dm=0.35)
table, impact, monthly = run(PROJECT_ID, FIELDS, TARGET_YEAR, BASELINE_YEAR, out_dir='outputs_feed_035', assumptions=a)
```
